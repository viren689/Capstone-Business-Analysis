# Customer Churn Analysis & Retention Strategy

## 1. Project Overview

Customer churn is an important business challenge because losing existing customers can reduce recurring revenue and increase the cost of acquiring new customers.

This project analyzes customer tenure, billing information, contract type, payment method, paperless billing, and demographic characteristics to identify patterns associated with customer churn.

The goal is to convert these findings into practical customer retention recommendations.

---

## 2. Business Problem

The business wants to understand why customers are leaving and identify customer segments that may have a higher risk of churn.

A better understanding of churn patterns can help the business prioritize retention activities, improve customer experience, and reduce customer loss.

---

## 3. Project Objective

The primary objective is to identify the key factors associated with customer churn and develop data-driven recommendations that can support customer retention strategies.

---

## 4. Business Questions

The analysis will answer the following questions:

1. What percentage of customers have churned?
2. Does customer tenure affect churn?
3. Does monthly spending relate to churn?
4. Which contract types have the highest churn rate?
5. Which payment methods are associated with higher churn?
6. Does paperless billing relate to customer churn?
7. Do senior citizens have different churn patterns?
8. Which customer segments should be prioritized for retention?

---

## 5. Success Metrics

The project will evaluate:

- Overall customer churn rate
- Churn rate by contract type
- Churn rate by payment method
- Churn rate by tenure
- Average monthly charges for churned and retained customers
- Average total charges for churned and retained customers

---

## 6. Analysis Techniques

The project will use multiple analytical techniques:

- Descriptive statistical analysis
- Customer segmentation and group comparison
- Correlation analysis
- Hypothesis testing
- Churn-rate analysis

---

## 7. Expected Business Outcome

The final analysis will identify high-risk customer segments and potential churn drivers.

Based on these findings, an implementation plan will be developed to help the business prioritize retention campaigns and monitor their effectiveness.

# 12. Data Cleaning and Validation

The dataset contains no missing values or duplicate records. However, we still need to check whether numerical and categorical variables contain invalid or unexpected values.

This step ensures that the data is logically consistent before exploratory analysis.

```python
# Check the minimum and maximum values of numerical variables
numerical_columns = [
    "Tenure",
    "MonthlyCharges",
    "TotalCharges",
    "SeniorCitizen",
    "Churn"
]

range_check = df[numerical_columns].agg(["min", "max"]).T

range_check
```

```python
# Check that binary variables contain only valid 0/1 values
binary_columns = ["SeniorCitizen", "Churn"]

for column in binary_columns:
    invalid_values = set(df[column].unique()) - {0, 1}
    
    if not invalid_values:
        print(f"{column}: validation passed.")
    else:
        print(f"{column}: invalid values found -> {invalid_values}")
```

```python
# Check the categories used in categorical variables
categorical_columns = [
    "Contract",
    "PaymentMethod",
    "PaperlessBilling"
]

for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].value_counts())
```

```python
# Verify that every customer has a unique identifier
duplicate_customer_ids = df["CustomerID"].duplicated().sum()

print(f"Duplicate Customer IDs: {duplicate_customer_ids}")
```

```python
# Confirm that customer IDs follow the expected format
invalid_customer_ids = ~df["CustomerID"].str.match(r"^C\d{5}$")

print(f"Invalid Customer IDs: {invalid_customer_ids.sum()}")
```

```python
# Create a concise data-quality summary
quality_summary = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Missing Values",
        "Duplicate Rows",
        "Duplicate Customer IDs"
    ],
    "Result": [
        len(df),
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df["CustomerID"].duplicated().sum()
    ]
})

quality_summary
```
